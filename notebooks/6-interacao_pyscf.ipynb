## Cálculo para dímero de água

```python
import pyscf

# dímero de água
mol = pyscf.M(
    atom = [['H', (-0.6305, -0.8758, 0.2588)], 
            ['O', (-1.5614, -0.8255, 0.5368)], 
            ['H', (-1.9991, -1.5191, 0.0288)], 
            ['O', (1.2821, -0.7524, -0.1108)], 
            ['H', (1.5168, 0.1326, -0.4203)], 
            ['H', (1.7973, -0.8645, 0.6993)]],
    basis = 'aug-cc-pvdz')
```

## Cálculos de DFT-D

```python
from pyscf.dft import KS

mf = mol.KS()
mf.xc = 'pbe' # shorthand for pbe,pbe
mf.disp = 'd3bj' # corrected version of D3 dispersion with Becke-Johnson damping
mf.kernel()
```

Exportanto a Densidade eletrônica para visualização

```python
from pyscf.tools import cubegen # para exportar aquivos cube

cubegen.density(mol, 'dimer_h2o_den.cube', mf.make_rdm1())
```

## Cálculo para trímero

```python
mol = pyscf.M(
    atom = [['H', (8.8983, -2.2843, -0.1789)], 
            ['O', (7.9708, -1.9757, -0.1200)], 
            ['H', (7.5408, -2.5819, 0.4951)], 
            ['O', (10.7652, -1.9603, -0.2644)], 
            ['H', (10.5807, -1.0101, -0.1202)], 
            ['H', (11.2830, -2.2241, 0.5066)], 
            ['O', (9.3715, 0.4200, 0.2688)], 
            ['H', (8.6226, -0.2038, 0.1882)], 
            ['H', (9.1408, 1.1658, -0.2985)]],
    basis = 'aug-cc-pvdz')
```

```python
mf = mol.KS()
mf.xc = 'pbe' # shorthand for pbe,pbe
mf.disp = 'd3bj' # corrected version of D3 dispersion with Becke-Johnson damping
mf.kernel()
```

```python
cubegen.density(mol, 'trimer_h2o_den.cube', mf.make_rdm1())
```
