## Cálculo para dímero de água

```python
import pyscf

# dímero de água
mol = pyscf.M(
    atom = [['H', (-0.6305, -0.8758, 0.2588)], 
            ['O', (-1.5614, -0.8255, 0.5368)], 
            ['H', (-1.9991, -1.5191, 0.0288)], 
            ['O', (1.2821, -0.7524, -0.1108)], 
            ['H', (1.5168, 0.1326, -0.4203)], 
            ['H', (1.7973, -0.8645, 0.6993)]],
    basis = 'aug-cc-pvdz')
```

## Cálculo de Energia sem correção D

```python
from pyscf.dft import KS

mf = mol.KS()
mf.xc = 'b3lyp' # B3LYP
mf.kernel()

Edft = mf.e_tot

print(f"Energia total (DFT): {Edft:.6f} Hartree")
```

## Cálculo de Energia com DFT-D

```python
mf.disp = 'd3bj' # correção D3 de dispersão com amortecimento de Becke-Johnson
mf.kernel()

Edftd = mf.e_tot

print(f"Energia total (DFT-D): {Edftd:.6f} Hartree")
```

## Correção D (dispersiva)

```python
print(f"Correção dispersiva da energia: {(Edftd-Edft):.6f} Hartree")
```

Exportanto a Densidade eletrônica para visualização

```python
from pyscf.tools import cubegen # para exportar aquivos cube

cubegen.density(mol, 'dimer_h2o_den.cube', mf.make_rdm1())
```
